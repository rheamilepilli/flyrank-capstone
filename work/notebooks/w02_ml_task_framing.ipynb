```python
import numpy as np
import pandas as pd

# Create synthetic starter data representing your unit of analysis (1 row = 1 content_hash_id)
np.random.seed(42)
content_ids = [f"hash_{i:04d}" for i in range(1, 6)]

data = {
    "content_hash_id": content_ids,
    "avg_imp_recent": np.random.randint(500, 5000, size=5),
    "avg_clicks_recent": np.random.randint(50, 500, size=5),
    "avg_pos_recent": np.round(np.random.uniform(1.5, 15.0, size=5), 1),
    "avg_imp_base": np.random.randint(500, 5000, size=5),
    "avg_clicks_base": np.random.randint(50, 500, size=5),
    "avg_pos_base": np.round(np.random.uniform(1.5, 15.0, size=5), 1),
}

df_sample = pd.DataFrame(data)

# Calculate the target column (click_delta)
df_sample["click_delta"] = (
    df_sample["avg_clicks_recent"] - df_sample["avg_clicks_base"]
)

print("Unit of Analysis DataFrame Preview (1 row = 1 content_hash_id):")
display(df_sample)
```

# Week 2: ML Task Framing — Content Refresh Opportunity Scoring

## 1. My Lane as an ML Task
* **Task Type:** Continuous Regression / Opportunity Scoring Engine
* **Core Objective:** Predict future organic performance decay to rank content items that need immediate content refreshes.

## 2. Target or Proxy Column
* **Target Variable (y):** `click_delta` (Continuous float value representing change in average organic clicks: Clicks_recent - Clicks_baseline).
* **Proxy Meaning:** Negative deltas signal content decay and loss of visibility; positive deltas indicate growing or stable content trajectory.

## 3. Success Metric
* **Primary Metric:** Mean Absolute Error (MAE) evaluated against a Zero-Change Baseline.
* **Secondary Metric:** Coefficient of Determination (R²) to evaluate variance capture.

## 4. Unit of Analysis as a Real Dataframe
* **Unit of Analysis:** One Row = One Content Item (`content_hash_id`) aggregated over temporal windows.

## 5. Why ML Beats a Fixed Rule Here
* **Non-Linear Interactions:** Fixed heuristics fail because high-authority evergreen content does not decay at a fixed calendar rate.
* **Multi-Variable Trajectory:** Machine learning models evaluate non-linear feature combinations to flag decay before traffic loss occurs.
* **Actionable Output:** Drives an automated priority queue for editorial teams, identifying which articles need immediate updates.
